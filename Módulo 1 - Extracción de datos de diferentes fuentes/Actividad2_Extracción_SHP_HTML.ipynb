## 1.- Análisis espacial de los municipios de Jalisco

Descarga una capa de municipios de Jalisco desde el INEGI o desde el portal de datos abiertos de Jalisco y guárdala en una carpeta local. El archivo principal debe tener extensión `.shp` y sus archivos auxiliares deben permanecer en la misma carpeta.

Puedes descargar capas reales desde estas fuentes:

1. **INEGI, Marco Geoestadístico:** [https://www.inegi.org.mx/temas/mg/](https://www.inegi.org.mx/temas/mg/). Descarga el Marco Geoestadístico y selecciona las capas de entidades, municipios o localidades. Conserva juntos los archivos `.shp`, `.shx`, `.dbf` y `.prj`.
2. **Datos abiertos de Jalisco:** [https://datos.jalisco.gob.mx/](https://datos.jalisco.gob.mx/). Busca conjuntos relacionados con municipios, localidades, infraestructura, salud, educación o territorio y revisa si ofrecen formato Shapefile.
3. **INEGI, mapas y datos geográficos:** [https://www.inegi.org.mx/](https://www.inegi.org.mx/). Utiliza el buscador con términos como `Jalisco municipios shapefile` o `Jalisco localidades`.

Leer una capa geográfica con GeoPandas, inspeccionar sus atributos y geometrías, construir un mapa temático y exportar resultados tabulares para su análisis posterior.

#### Instrucciones

1. Carga la capa con `gpd.read_file()`.
2. Inspecciona `head()`, `columns`, `shape`, `crs` y `geometry.geom_type`.
3. Identifica la columna que contiene el nombre o clave del municipio.
4. Comprueba si existen valores faltantes, geometrías vacías o geometrías inválidas.
5. Calcula el área de cada municipio. Antes de calcularla, reproyecta la capa a un CRS proyectado apropiado para Jalisco.
6. Ordena los municipios por superficie y muestra los cinco más grandes.
7. Crea un mapa de los municipios utilizando `plot()`, bordes visibles y una escala de color basada en el área.
8. Exporta una tabla sin geometría a `municipios_jalisco.csv`.
9. Exporta la capa procesada a GeoPackage o Shapefile.

#### Contesta las siguientes Preguntas

1. ¿Cuántos municipios contiene la capa?
2. ¿Qué sistema de coordenadas utiliza originalmente?
3. ¿Qué tipo de geometría tiene la capa?
4. ¿Cuál es el municipio con mayor superficie?
5. ¿Cuál es el municipio con menor superficie?
6. ¿Hay geometrías inválidas o atributos faltantes?
7. ¿Por qué no es adecuado calcular áreas directamente con un CRS geográfico en grados?
8. ¿Qué ventajas tiene exportar los atributos a CSV y conservar la geometría en un formato geoespacial?



## 2.- Extraer información de una página web

Consulta la página [Programming Historian](http://programminghistorian.org/es/lecciones/) y realiza lo siguiente:

1. Descarga su contenido con `requests`.
2. Analiza el HTML con `BeautifulSoup`.
3. Extrae de cada lección el título, autor, resumen, enlace, fecha de publicación, tópicos, actividad y dificultad.
4. Guarda la información en un `DataFrame` y exporta el resultado como `lecciones-ph_es.csv`.
